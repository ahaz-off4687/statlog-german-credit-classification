```python

from ucimlrepo import fetch_ucirepo
import pandas as pd
import numpy as np

# Fetch dataset
statlog_german_credit_data = fetch_ucirepo(id=144) 

# Separate features and target
X = statlog_german_credit_data.data.features 
y = statlog_german_credit_data.data.targets 

# Combine features and target
df = pd.concat([X, y], axis=1)

# Display dataset information
print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst five records:")
display(df.head())

print("\nFeature names:")
print(X.columns.tolist())

print("\nTarget variable:")
print(y.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())

print("\nTarget distribution:")
print(y.value_counts())

print("\nData types:")
print(df.dtypes)
```

```python


from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

# Convert target dataframe into a Series
y = y.squeeze()

# Identify categorical and numerical columns
categorical_columns = X.select_dtypes(include=["object"]).columns.tolist()

numerical_columns = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("Categorical features:", len(categorical_columns))
print("Numerical features:", len(numerical_columns))

# Split the dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)

# Create One-Hot Encoder
try:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )

# Create preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", encoder, categorical_columns),
        ("numerical", StandardScaler(), numerical_columns)
    ]
)

# Apply preprocessing to training and testing data
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("\nPreprocessing completed successfully!")

print("Processed training data shape:",
      X_train_processed.shape)

print("Processed testing data shape:",
      X_test_processed.shape)
```

```python

# Import required libraries

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

# Create a function for preprocessing

def create_preprocessor():

    try:
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    except TypeError:
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )

    preprocessor = ColumnTransformer(
        transformers=[
            ("categorical", encoder, categorical_columns),
            ("numerical", StandardScaler(), numerical_columns)
        ]
    )

    return preprocessor


# Create the five classification models

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=5
    ),

    "Gaussian Naive Bayes": GaussianNB(),

    "SVM": SVC(
        probability=True,
        random_state=42
    )
}


# Create pipelines for all five models

model_pipelines = {}

for model_name, model in models.items():

    pipeline = Pipeline([
        ("preprocessor", create_preprocessor()),
        ("classifier", model)
    ])

    model_pipelines[model_name] = pipeline


# Train all five models

for model_name, pipeline in model_pipelines.items():

    pipeline.fit(X_train, y_train)

    print(model_name, "trained successfully!")


print("\nAll five classification models have been trained.")
```

```python

# Import evaluation metrics

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report
)

# Create an empty list to store model results

results = []

# Evaluate each trained model

for model_name, pipeline in model_pipelines.items():

    # Predict the test data
    y_pred = pipeline.predict(X_test)

    # Get prediction probabilities
    y_prob = pipeline.predict_proba(X_test)

    # Find the column corresponding to class 2
    class_index = list(pipeline.classes_).index(2)

    # Calculate AUC for class 2
    auc = roc_auc_score(
        (y_test == 2).astype(int),
        y_prob[:, class_index]
    )

    # Calculate performance metrics
    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test, y_pred,
        pos_label=2,
        zero_division=0
    )

    recall = recall_score(
        y_test, y_pred,
        pos_label=2,
        zero_division=0
    )

    f1 = f1_score(
        y_test, y_pred,
        pos_label=2,
        zero_division=0
    )

    mcc = matthews_corrcoef(y_test, y_pred)

    # Store the results
    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "AUC": auc,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "MCC": mcc
    })


# Convert results into a dataframe

results_df = pd.DataFrame(results)

# Display comparison table

print("Classification Model Performance Comparison")
print(results_df.round(4).to_string(index=False))
```

```python

from sklearn.metrics import confusion_matrix, classification_report

# Evaluate each model using the test data

for model_name, pipeline in model_pipelines.items():

    # Predict the test records
    y_pred = pipeline.predict(X_test)

    # Generate confusion matrix
    cm = confusion_matrix(y_test, y_pred, labels=[1, 2])

    print("\n----------------------------------")
    print("Model:", model_name)
    print("----------------------------------")

    print("\nConfusion Matrix:")
    print(cm)

    print("\nClassification Report:")
    print(classification_report(
        y_test,
        y_pred,
        labels=[1, 2],
        zero_division=0
    ))

# Identify the model with the highest MCC

best_model_name = results_df.loc[
    results_df["MCC"].idxmax(), "Model"
]

print("\nBest model based on MCC:", best_model_name)
```

```python

# Import required libraries

import os
import joblib
import pandas as pd

# Create a folder to store trained models

os.makedirs("saved_models", exist_ok=True)

# Save all five trained models

for model_name, pipeline in model_pipelines.items():

    # Convert model name into a file name
    file_name = model_name.lower().replace(" ", "_")

    # Save the complete pipeline
    joblib.dump(
        pipeline,
        "saved_models/" + file_name + ".pkl"
    )

    print(model_name, "saved successfully!")


# Save the test dataset with actual target values

test_data = X_test.copy()
test_data["class"] = y_test

test_data.to_csv("test_data.csv", index=False)

print("\nTest dataset saved successfully!")


# Save the model comparison results

results_df.to_csv("model_comparison.csv", index=False)

print("Model comparison results saved successfully!")


# Display saved files

print("\nSaved model files:")

for file in os.listdir("saved_models"):
    print(file)
```
